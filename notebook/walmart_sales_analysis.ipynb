### data cleaning and exploration ###

```python
import matplotlib.pyplot as plt
from sqlalchemy import create_engine # works as connector
import pymysql
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('../data/Walmart.csv', encoding_errors='ignore')
```

```python
df.head(10)
df.shape
df.info()
df.describe()
df.isnull().sum()
```

```python
# all duplicated values

df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
# dropping all rows with missing records 

df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
# converting the unit_price column to numeric by removing the $ sign and converting to float

df['unit_price'] = df['unit_price'].str.replace('$' , '').astype(float)
```

```python
df.columns
```

```python
df['total'] = df['quantity'] * df['unit_price']
```

```python
df.head()
```

```python
df.to_csv('Walmart_cleaned.csv', index=False)
```

```python
# my sql comnection 

from sqlalchemy import create_engine

engine_mysql = create_engine(
    'mysql+pymysql://root:Root%40123@localhost:3306/walmart_db'
)

try:
    connection = engine_mysql.connect()
    print("Connection to MySQL database successful!")
    connection.close()
except Exception as e:
    print(f"Error connecting to MySQL database: {e}")
```

```python
df.to_sql('Walmart_cleaned', con=engine_mysql, if_exists='append', index=False)
```
